## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def calcular_plantas(n):
    if n == 0:
        return 0
    if n == 1:
        return 1

    memo = {0: 0, 1: 1}

    for i in range(2, n + 1):
        memo[i] = memo[i - 1] + memo[i - 2]

    return memo[n]

while True:
    try:
        mes = int(input("Ingrese el mes: "))

        if mes < 0:
            print("Por favor, ingrese un número igual o mayor a 0.")
            continue

        resultado = calcular_plantas(mes)
        print(f"La produccion del mes {mes} es: {resultado}")
        break

    except ValueError:
        print("Error: Ingreso inválido. Por favor, escriba únicamente números enteros.")
```
